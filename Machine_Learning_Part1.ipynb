```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

import warnings
warnings.filterwarnings('ignore')


# =========================
# 1. LOAD DATA
# =========================

df = pd.read_csv('insurance.csv')


# =========================
# 2. BASIC EDA
# =========================

print(df.head())
print(df.shape)
print(df.columns)
print(df.dtypes)
print(df.info())
print(df.describe())
print(df.describe(include='object'))


# =========================
# 3. MISSING VALUES
# =========================

print("\nMissing Values:")
print(df.isnull().sum())

print("\nMissing Percentage:")
print((df.isnull().sum() / len(df)) * 100)


# =========================
# 4. DUPLICATES
# =========================

print("\nDuplicates:")
print(df.duplicated().sum())

df_cleaned = df.drop_duplicates().copy()


# =========================
# 5. COLUMN TYPES
# =========================

numeric_columns = df_cleaned.select_dtypes(
    include=['int64', 'float64']
).columns

categorical_columns = df_cleaned.select_dtypes(
    include=['object']
).columns

print("\nNumerical Columns:")
print(numeric_columns)

print("\nCategorical Columns:")
print(categorical_columns)


# =========================
# 6. NUMERICAL DISTRIBUTIONS
# =========================

for col in numeric_columns:
    plt.figure(figsize=(7, 3))

    sns.histplot(
        df_cleaned[col],
        kde=True,
        bins=20
    )

    plt.title(f'Distribution of {col}')
    plt.show()


# =========================
# 7. SKEWNESS
# =========================

print("\nSkewness:")
print(df_cleaned[numeric_columns].skew())


# =========================
# 8. BOXPLOTS / OUTLIERS
# =========================

for col in numeric_columns:
    plt.figure(figsize=(7, 2))

    sns.boxplot(
        x=df_cleaned[col]
    )

    plt.title(f'Boxplot of {col}')
    plt.show()


# =========================
# 9. CATEGORICAL ANALYSIS
# =========================

for col in categorical_columns:

    print(f"\nValue Counts for {col}:")
    print(df_cleaned[col].value_counts())

    plt.figure(figsize=(6, 3))

    sns.countplot(
        x=col,
        data=df_cleaned
    )

    plt.title(f'Count of {col}')
    plt.show()


# =========================
# 10. CORRELATION
# =========================

plt.figure(figsize=(8, 5))

sns.heatmap(
    df_cleaned[numeric_columns].corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap')
plt.show()


# =========================
# 11. RELATIONSHIPS
# =========================

sns.scatterplot(
    x='age',
    y='charges',
    data=df_cleaned
)

plt.title('Age vs Charges')
plt.show()


sns.scatterplot(
    x='bmi',
    y='charges',
    data=df_cleaned
)

plt.title('BMI vs Charges')
plt.show()


sns.boxplot(
    x='smoker',
    y='charges',
    data=df_cleaned
)

plt.title('Smoker vs Charges')
plt.show()


# =========================
# 12. FEATURE ENGINEERING
# =========================

df_cleaned['bmi_category'] = pd.cut(
    df_cleaned['bmi'],
    bins=[
        0,
        18.5,
        24.9,
        29.9,
        float('inf')
    ],
    labels=[
        'Underweight',
        'Normal',
        'Overweight',
        'Obese'
    ]
)


df_cleaned['age_group'] = pd.cut(
    df_cleaned['age'],
    bins=[
        0,
        25,
        40,
        60,
        float('inf')
    ],
    labels=[
        'Young',
        'Adult',
        'Middle_Age',
        'Senior'
    ]
)


df_cleaned['has_children'] = (
    df_cleaned['children'] > 0
).astype(int)


df_cleaned['obese_smoker'] = (
    (df_cleaned['bmi'] >= 30) &
    (df_cleaned['smoker'] == 'yes')
).astype(int)


df_cleaned['age_bmi'] = (
    df_cleaned['age'] *
    df_cleaned['bmi']
)


# =========================
# 13. LOG TRANSFORMATION
# =========================

df_cleaned['log_charges'] = np.log1p(
    df_cleaned['charges']
)


# =========================
# 14. ONE-HOT ENCODING
# =========================

df_encoded = pd.get_dummies(
    df_cleaned,
    columns=[
        'sex',
        'smoker',
        'region',
        'bmi_category',
        'age_group'
    ],
    drop_first=True,
    dtype=int
)


# =========================
# 15. FEATURES & TARGET
# =========================

X = df_encoded.drop(
    columns=[
        'charges',
        'log_charges'
    ]
)

y = df_encoded['charges']


# =========================
# 16. TRAIN TEST SPLIT
# =========================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)


# =========================
# 17. FEATURE SCALING
# =========================

scale_columns = [
    'age',
    'bmi',
    'children',
    'age_bmi'
]

scaler = StandardScaler()

X_train[scale_columns] = scaler.fit_transform(
    X_train[scale_columns]
)

X_test[scale_columns] = scaler.transform(
    X_test[scale_columns]
)


# =========================
# 18. FINAL OUTPUT
# =========================

print("\nX Train Shape:")
print(X_train.shape)

print("\nX Test Shape:")
print(X_test.shape)

print("\nFinal Training Data:")
print(X_train.head())
```
